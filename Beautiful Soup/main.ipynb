# BeautifulSoup
With this library, what we do is generate HTTP GET and POST requests to obtain information from a specific web page.

When using BeautifulSoup, the best way to make requests is with an XML file created with the help of the **lxml** library.

The different and most common tags are the following:

```html
<a> for hyperlink
<table> for tables
<tr> for table rows
<th> for table headers
<td> for table cells
```

## Introduction and Single Element Extraction
To start using BeautifulSoup, we must import the BeautifulSoup and Requests modules.
In html_text, we use text to return the HTML code. If we don't use the text method, it will textually return the XML connection status, whether it is 200 (successful) or another number (failed).

#Scraping the dia page (single product)

````python
from bs4 import BeautifulSoup
from requests_html import HTMLSession

session = HTMLSession()

html_text = session.get(
    "https://www.dia.es/charcuteria-y-quesos/jamon-cocido-lacon-fiambres-y-mortadela/c/L2001",
).text
soup = BeautifulSoup(html_text, "lxml")

# What we do is first put the main object followed by the class
# Instead of using children we can search for an object inside another object (for now)

job = soup.find("li", class_="product-card-list__item-container")

if job:
    print(job.text)

    # In the following line we use the method "text". It is quite similar to the property innertext in HTML

    product = job.find("p", class_="search-product-card__product-name").text.strip("")  # type: ignore

    print(product)
````

Instead of finding an element by class type, we could also find it by **id** or **tag name**. We could do it in the following way:

## Web scraping of multiple elements
We have already seen how to extract a single element. Now it's time to see how to extract several. The following code shows how we extract multiple elements.

---


````python
from bs4 import BeautifulSoup
from requests_html import HTMLSession

session = HTMLSession()

html_text = session.get(
    "https://www.dia.es/charcuteria-y-quesos/jamon-cocido-lacon-fiambres-y-mortadela/c/L2001",
).text
soup = BeautifulSoup(html_text, "lxml")

# What we do is first put the main object followed by the class
# Instead of using children we can search for an object inside another object (for now)

jobs = soup.find_all("li", class_="product-card-list__item-container")
errors = []

for job in jobs:
    try:
        print(f"JOB TEXT\n{job.text}")

        product_element = job.find("p", class_="search-product-card__product-name")

        # if product_element:
        product = product_element.text.strip("")
        print(f"PRODUCT\n{product}")
    except AttributeError as e:
        errors.append(e)

# [f"Errors: {error}" for error in errors]
````

The “keyword” argument allows you to select tags that contain a particular attribute or a series of attributes:


## Objects and Navigation

There are 2 types of objects in BeautifulSoup:
- BeautifulSoup Object: Objects related with the **find** or **find_all** methods.
- Tag Object: Objects that represent a single HTML or XML tag within the document. Example: `bs.div.h1`, which represents an `<h1>` tag inside a `<div>` tag.

### BeautifulSoup Objects

The **find** and **find_all** methods are more intended to extract values representing classes.
**find** is used to extract the first and `ONLY` piece of information from the text returned using **BeautifulSoup** on the HTML code extracted with **requests_html** or **requests**.

Example:

````python
from bs4 import BeautifulSoup
from requests_html import HTMLSession

session = HTMLSession()

response = session.get(
    "https://www.dia.es/charcuteria-y-quesos/jamon-cocido-lacon-fiambres-y-mortadela/c/L2001",
    timeout=None,
)  # pyright: ignore[reportGeneralTypeIssues]

soup = BeautifulSoup(response.text, "lxml")

# Change class and label

products_containers = soup.find_all("li", class_="product-card-list__item-container")
products = []

for idx, container in enumerate(products_containers):
    name = container.find("p", class_="search-product-card__product-name")
    price = container.find("p", class_="search-product-card__active-price")
    price_per_unit = container.find("p", class_="search-product-card__price-per-unit")

    if not name or not price or not price_per_unit:
        continue

    products.append({
        "#"             : idx + 1,
        "name"          : name.text.strip(),
        "price"         : price.text.replace("\xa0", ""),
        "price_per_unit": price_per_unit.text.replace("\xa0", "").strip("() "),
    })

# products_name = [
#     container.find("p", class_="search-product-card__product-name").text.strip()
#     for container in products_containers
#     if container.find("p", class_="search-product-card__product-name")
# ]

# products_price = [
#     (
#         container.find("p", class_="search-product-card__active-price").text.replace(
#             "\xa0",
#             "",
#         ),
#         container.find("p", class_="search-product-card__price-per-unit")
#         .text.replace("\xa0", "")
#         .replace("(", "")
#         .replace(")", "")
#         .strip(),
#     )
#     for container in products_containers
#     if container.find("p", class_="search-product-card__active-price")
#     and container.find("p", class_="search-product-card__price-per-unit")
# ]

# products = [
#     {
#         "#": product + 1,
#         "name": products_name[product],
#         "price": products_price[product][0],
#         "price_per_unit": products_price[product][1],
#     }
#     for product in range(len(products_name))
# ]

products
````

````python
import pandas as pd

df = pd.DataFrame(products, index=[product["#"] for product in products])

df = df.drop(columns=["#"])
df = df.rename(
    columns={
        "name"          : "Product Name",
        "price"         : "Price",
        "price_per_unit": "Price per Unit",
    },
)

df
````

### Tag Objects

To access to the information of these objects after using BFS, we can search into the children tags one by one returning the text of each one.

Example:

Inside the class `col-xs-6 col-sm-4 col-md-3 col-lg-3` with the label `li`.

Inside we have several children tags, `h3`, after `a`. Like this:

<img src="../assets/img/image2.png" alt="Inspeccionar Elemento" width="600" />



The code should be like this (using **text** to extract the text):

````python
response = session.get("https://books.toscrape.com/")

bs = BeautifulSoup(response.text, "lxml")
soup = bs.find_all("li", class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")

for text in soup:
    title = text.h3.a.text
    print(title)
````

**There are a few times, with the use of JS, the represented information is not inside the element but inside the class itself**:

<img src="../assets/img/image3.png" alt="Information inside the class itself" width="1000" />

In this case, we have to access to the class name and then to access to the information with need to treat it as a dictionary:

````python
bs = BeautifulSoup(response.text, "lxml")
soup = bs.find_all("li", class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")

for text in soup:
    stars = text.p["class"][1]
    print(stars)
````

This will allow us to access the second object that represents the class.

In case we want to access to the attrs in the tag (href, src or img) we must use the property **attrs**.

````python
bs = BeautifulSoup(response.text, "lxml")
soup = bs.find_all("li", class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")

for text in soup:
    title = text.h3.a.attrs
    print(title)

    # To access to the href attribute we can do:

    href = title["href"]
    print(href, "\n")
````

### Siblings

A comfortable way to grab table content is using siblings, which is practically designed for scraping tables. Using the previous table as an example, we do the following to obtain the values:

````python
from typing import cast
from urllib.request import urlopen

from bs4 import BeautifulSoup, Tag

html = urlopen("http://www.pythonscraping.com/pages/page3.html")
bs = BeautifulSoup(html, "lxml")

table = bs.find("table", {"id": "giftList"})
table = cast("Tag", table)

if table.tr:
    for siblings in table.tr.next_siblings:
        print(siblings)
````

### Accessing Attributes
HTML tags often handle multiple objects. That is to say, tags like `<a>` or `<p>` contain more objects besides the text. To access them we need to use the `attrs` property.

**For example, if we want to get the attributes names (href, src, etc.) found in the tag, we use the following code:**

`my_tag.attrs`

If there are multiple attributes, we can access them like a dictionary:

`my_tag.attrs['attribute_name']`

### Lambda Functions

We can use the lambda function as a keyword argument in the find_all.

Example for extracting all the tags with only 2 attrs:

```html
<div class="body" id="content"></div>
<span style="color: red;"></span>
```

```python
bs.find_all(lambda tag: len(tag.attrs) == 2)
```

### Downloading Files

To download files with urllib, we can use the following code:

````python
from pathlib import Path

import urllib3


def download_file(url: str, file_name: str) -> None:
    http = urllib3.PoolManager()

    with http.request("GET", url, preload_content=False) as r:
        Path(file_name).write_bytes(r.data)

    print(f"File '{file_name}' downloaded successfully.")


url_file = "https://www.dia.es/product_images/273961/273961_ISO_0_ES.jpg?imwidth=176"
local_file_name = "/home/fenyxz/CODING/WORK/Scraping/Scraping_Libraries/Beautiful Soup/jamon_cocido_lacon.jpg"

download_file(url_file, local_file_name)
````

## Mechanical Soup

The mechanicalsoup is a Python library for Python interactions automatization for websites.\
It is built on top of the `Requests` library and `BeautifulSoup`. It provides a simple API to automate tasks such as filling out forms, clicking buttons, and navigating through web pages.

For example, we will automatize a search using the Google explorer and search for *Messi*:

````python
import mechanicalsoup

headers = {
    "User-Agent"     : "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/100.0.4896.127 Safari/537.36",
    "Accept"         : "text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,image/apng,*/*;q=0.8",
    "Accept-Language": "es-ES,es;q=0.9",
    "Connection"     : "keep-alive",
    "Referrer-Policy": "origin-when-cross-origin"
}

browser = mechanicalsoup.StatefulBrowser()
browser.session.headers = headers
browser.open("https://www.google.com/")

browser.select_form("[action='/search']")
browser["q"] = "Messi"
browser.submit_selected()

for link in browser.links():
    print(link.text, link.attrs["href"])

# In this case we want to obtain the HTML code of the website after the search, so we can do it with the following code:

print(browser.get_current_page().body)

# Basically we can use this library to submit forms, click buttons and navigate through the website. It is a very useful library for web scraping when we need to interact with the website.
````

````python
import mechanicalsoup

browser = mechanicalsoup.StatefulBrowser()
browser.open("https://html.duckduckgo.com/html/")

browser.select_form(selector="[action='/html/']")
browser["q"] = "MechanicalSoup"

browser.submit_selected(btnName="btnG")

print(browser.get_current_page())

for link in browser.links():
    print(f"{link.text} -< {link.attrs["href"]}")
````
