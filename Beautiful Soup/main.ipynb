## BeautifulSoup
With this library, what we do is generate HTTP GET and POST requests to obtain information from a specific web page.

When using BeautifulSoup, the best way to make requests is with an XML file created with the help of the **lxml** library.

The different and most common tags are the following:

```html
<a> for hyperlink
<table> for tables
<tr> for table rows
<th> for table headers
<td> for table cells
```

#### Introduction and web scraping of an element
To start using BeautifulSoup, we must import the BeautifulSoup and Requests modules.
In html_text, we use text to return the HTML code. If we don't use the text method, it will textually return the XML connection status, whether it is 200 (successful) or another number (failed).

---
#Scraping the dia page (single product)

````python
from bs4 import BeautifulSoup
from requests_html import HTMLSession

session = HTMLSession()

html_text = session.get(
    "https://www.dia.es/charcuteria-y-quesos/jamon-cocido-lacon-fiambres-y-mortadela/c/L2001",
).text
soup = BeautifulSoup(html_text, "lxml")

# What we do is first put the main object followed by the class
# Instead of using children we can search for an object inside another object (for now)

job = soup.find("li", class_="product-card-list__item-container")

if job:
    print(job.text)

    # In the following line we use the method "text". It is quite similar to the property innertext in HTML

    product = job.find("p", class_="search-product-card__product-name").text.strip("")  # type: ignore

    print(product)
````

Instead of finding an element by class type, we could also find it by **id** or **tagname**. We could do it in the following way:

```python
# ID

# Tagname

```

To finally obtain the text of a single final object element from which we extract the information, we use **text**.

#### Web scraping of multiple elements
We have already seen how to extract a single element. Now it's time to see how to extract several. The following code shows how we extract multiple elements.

---


````python
from bs4 import BeautifulSoup
from requests_html import HTMLSession

session = HTMLSession()

html_text = session.get(
    "https://www.dia.es/charcuteria-y-quesos/jamon-cocido-lacon-fiambres-y-mortadela/c/L2001",
).text
soup = BeautifulSoup(html_text, "lxml")

# What we do is first put the main object followed by the class
# Instead of using children we can search for an object inside another object (for now)

jobs = soup.find_all("li", class_="product-card-list__item-container")
errors = []

for job in jobs:
    try:
        print(f"JOB TEXT\n{job.text}")

        product_element = job.find("p", class_="search-product-card__product-name")

        # if product_element:
        product = product_element.text.strip("")
        print(f"PRODUCT\n{product}")
    except AttributeError as e:
        errors.append(e)

# [f"Errors: {error}" for error in errors]
````

Another way to get text with find_all for class-type objects would be the following:
We have the following example:

```html
<span class="red">
    Heavens! what a virulent attack!
</span>
```

To extract the information we use the following code:

```python
name_list = bs.find_all("span", class_="green")

```

The parameters found in find and find_all are the following:

```python
find_all(tag, attrs, recursive, string, limit, **kwargs)

find(tag, attrs, recursive, string, **kwargs)
```

You can pass a list of tags in the first parameter so that it searches for all the tags in the list:

```python
find_all(["h1", "h2", "h3", "h4", "h5", "h6"])
```

As we explained earlier, first we put the tag name and then the attribute:

```python
find_all("span", {"class": {"green", "red"}})
```

Using the parameter “text=” we can extract all objects which have the name followed by the string. Suppose we want to extract all objects with the name “the prince”, we would do it as follows:

```python
name_list = bs.find_all(text="the prince")
```

The “keyword” argument allows you to select tags that contain a particular attribute or a series of attributes:

```python
title = bs.find_all(id="title", {"class":"text"})
```

If we only want to find the first object we find, we will use only find:

```python
title = bs.find(id="title")
```

The keyword argument can be useful in some situations. However, it is technically redundant as a feature of BeautifulSoup.
Keep in mind that anything that can be done with keywords can also be achieved using techniques discussed later in this chapter
(consult regular_express and lambda_express)

Example, the next two lines are equivalent:

---

```python
bs.find_all(class_="external-link")
bs.find_all("", {"class": "external-link"})
```

---

**If we want to search for a class directly without having to specify the tag, we can do so in the next way:**

```python
bs.find_all(class_="external-link")
```

**Or**

```python
bs.find_all("", {"class": "external-link"})
```

#### Another objects in BeautifulSoup

There are 2 types of objects in BeautifulSoup:
- BeautifulSoup Object: Objects related with the **find** or **find_all** methods.
- Tag Object: Objects that represent a single HTML or XML tag within the document. Example: `bs.div.h1`, which represents an `<h1>` tag inside a `<div>` tag.

##### Objects labels/tags

To access to the information of this objects after using BFS, we can search into the children tags one by one returning the text of each one.

Example:

Inside the class `col-xs-6 col-sm-4 col-md-3 col-lg-3` with the label `li`.

Inside we have several children tags, `h3`, after `a`. Like this:

<img src="../assets/img/image2.png" alt="Inspeccionar Elemento" width="600" />

The code should be like this (using **text** to extract the text):

````python
response = session.get("https://books.toscrape.com/")

bs = BeautifulSoup(response.text, "lxml")
soup = bs.find_all("li", class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")

for text in soup:
    title = text.h3.a.text
    print(title)
````

**There are a few times, with the use of JS, the represented information is not inside the element but inside the class itself**:

<img src="../assets/img/image3.png" alt="Information inside the class itself" width="1000" />

In this case, we have to access to the class name and then to access to the information with need to treat it as a dictionary:

````python
bs = BeautifulSoup(response.text, "lxml")
soup = bs.find_all("li", class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")

for text in soup:
    stars = text.p["class"][1]
    print(stars)
````

This will allow us to access the second object that represents the class.

In case we want to access to the attrs in the tag (href, src or img) we must use the property **attrs**.

````python
bs = BeautifulSoup(response.text, "lxml")
soup = bs.find_all("li", class_="col-xs-6 col-sm-4 col-md-3 col-lg-3")

for text in soup:
    title = text.h3.a.attrs
    print(title)

    # To access to the href attribute we can do:

    href = title["href"]
    print(href, "\n")
````

To get all the elements of the table we would use the following code:

````python
from urllib.request import urlopen

from bs4 import BeautifulSoup

html = urlopen("http://www.pythonscraping.com/pages/page3.html")
bs = BeautifulSoup(html, "lxml")
table = bs.find("table", {"id": "giftList"})

for child in table.children:
    print(child)

# Is better to use children() instead of descendants(). Is more selected.
````

#### Siblings

A comfortable way to grab table content is using siblings, which is practically designed for scraping tables. Using the previous table as an example, we do the following to obtain the values:

````python
from typing import cast
from urllib.request import urlopen

from bs4 import BeautifulSoup, Tag

html = urlopen("http://www.pythonscraping.com/pages/page3.html")
bs = BeautifulSoup(html, "lxml")

table = bs.find("table", {"id": "giftList"})
table = cast("Tag", table)

if table.tr:
    for siblings in table.tr.next_siblings:
        print(siblings)
````

#### Parents

In case we want to get the value of the parent tag of a specific tag, we can use:

````python
img = bs.find("img", {"src": "../img/gifts/img1.jpg"})

if img and img.parent and img.parent.previous_sibling:
    print(img.parent.previous_sibling.get_text())
````

In this code we access to the img tag, then to its parent and finally to the previous sibling of the parent, which is a td tag. With get_text() we obtain the text inside that td tag.

### Accessing attributes
HTML tags often handle multiple objects. That is to say, tags like `<a>` or `<p>` contain more objects besides the text. To access them we need to use the `attrs` property.

**For example, if we want to get the attributes names (href, src, etc) found in the tag, we use the following code:**

`my_tag.attrs`

If there are multiple attributes, we can access them like a dictionary:

`my_tag.attrs['attribute_name']`

### Lambda Function

We can use the lambda function as a keyword argument in the find_all.

Example for extracting all the tags with only 2 attrs:

```html
<div class="body" id="content"></div>
<span style="color: red;"></span>
```

```python
bs.find_all(lambda tag: len(tag.attrs) == 2)
```
