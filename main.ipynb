### Requests
#### Metodo POST

El método POST es un tipo de solicitud http que consiste en enviar una solicitud acompañada de parametros a un formulario web en lugar de un enlace como sería las solicitudes tipo get.

Vamos a hacer una solictud http tipo pero la haremos a la página de investing:

##### Solicitud con párametros

En el 100% de los casos los casos en las solicitudes tipo POST intervienen mínimo 2 atributos. En total son 3 los cuales sería los siguientes:

```requests.post(‘<URL>’, data = <object>, json = <objeto_diccionario>)```

En data y json llevan diccionarios. La diferencia es que en json los datos deben de ser antes convertido a json con la librería json antes de establecerse.

Aquí tenemos un ejemplo:

````python
import requests

URL = "https://httpbin.org/post"

data_dictionary = {
    "name": "username",
    "password": "password123",
    "email": "user@example.com",
}

response = requests.post(URL, data=data_dictionary)

print(response.text)
````

With the requests library, we can also use the HTMLSession class from the requests_html library (more secure and effective) to make POST requests.

Here's how to do it:

````python
from requests_html import HTMLSession

session = HTMLSession()

data_dictionary = {
    "name": "username",
    "password": "password123",
    "email": "user@example.com",
}

response = session.post(URL, data=data_dictionary)

print(response.text)
````

#### Encabezados

(Misma informacion del principio)

Cuando hacemos solicitudes http sea con requests,requests_html, urllib o httpx es conveniente de que establezcamos los encabezados que utilizaremos a la hora de hacer una solicitud http.

Los diferentes tipos de encabezados que tenemos son los siguientes:

```text
Host |https://www.google.com/

Connection| keep-alive

Accept|text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8

User-Agent |Mozilla/5.0 (Macintosh; Intel Mac OS X 10_9_5) AppleWebKit/537.36 (KHTML,like Gecko) Chrome/39.0.2171.95 Safari/537.36

Referer | https://www.google.com/

AcceptEncoding | gzip, deflate, sdch

AcceptLanguage |en-US,en;q=0.8
```

Y aquí están las cabezeras que salen por defecto al usar urllib:

```text
Accept-Encoding | identity

User-Agent |Python-urllib/3.4
```

Puedes modificar los encabezados de la siguiente manera:

```python
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64; rv:84.0) Gecko/20100101  Firefox',
    'Accept-Language': 'en-GB, en:q= 0.5',
    'Referer' : 'https://google.com',
    'DNT' : "1",
    "Arian": "food"
}
```

Aqui una breve explicacion de los encabezados anteriores:

1. 'Accept-Langue': 'en-GB, en:q=0.5' -> Este encabezado indica las preferencias de idioma del cliente. En este caso, el cliente prefiere el inglés británico (en-GB) con una ponderación de 0.5 y el inglés genérico (en) con una ponderación predeterminada. El servidor puede usar esta información para proporcionar contenido en el idioma preferido del usuario
2. 'Referer': 'https://google.com' -> El encabezado "Referer" (a menudo mal escrito como "Referrer") indica la página desde la cual el usuario llegó a la página actual. En este caso, el usuario llegó a esta página desde el sitio web "https://google.com". Este encabezado se utiliza para rastrear el origen de los visitantes y se usa en análisis web y seguimiento de referidos.
3. 'DNT': '1' -> El encabezado "Do Not Track" (DNT) es un valor que puede ser "1" o "0" (habilitado o deshabilitado). Cuando está habilitado ("1"), indica que el usuario no desea ser rastreado. Esto es una solicitud del usuario a los sitios web y servicios para que no recopilen información sobre su comportamiento de navegación. No todos los sitios web respetan esta preferencia.
4. 'Arian': 'food' -> Este encabezado es un ejemplo de un encabezado personalizado que no es estándar en HTTP. Los encabezados personalizados se pueden utilizar para enviar información específica del cliente o aplicaciones web. En este caso, el cliente está enviando un encabezado personalizado llamado "Arian" con el valor "food".

Otros encabezados que podriamos poner son los siguientes:

````python
import requests

from bs4 import BeautifulSoup

session = requests.Session()
headers = {
    "User-Agent": "Mozilla/5.0 (Macintosh; Intel Mac OS X 10_9_5)",
    "Accept": (
        "text/html,application/xhtml+xml,application/xml;q=0.9,image/webp,*/*;q=0.8"
    ),
}
url = (
    "https://www.whatismybrowser.com/developers/what-http-headers-is-my-browser-sending"
)
req = session.get(url, headers=headers)
bs = BeautifulSoup(req.text, "html.parser")
print(bs.find('table', {'class': 'table-striped'}).get_text)
````

#### Cookies

Las cookies son parámetros que se encuentran en nuestro navegador que tienen la finalidad de informar a las páginas web de los movimientos que llegamos a hacer, el tiempo que estamos conectados, etc. Las páginas web nos exigen acceptarlas y tenerlas alojadas en nuestro navegador para poder navegar. Muchos sitios web llegan a bloquerte por no hacer nada en el sentido de hacer alguna acción para protegerse de los bots.

Ahora lo que haremos serán que en formulario creado en http://pythonscraping.com/pages/cookies/login.html que que el objeto action se encuentra en http://pythonscraping.com/pages/cookies/welcome.php contiene el link a la pagina principal http://pythonscraping.com/pages/cookies/profile.php.

Si intenta acceder a la página de bienvenida o a la página de perfil sin iniciar sesión primero, recibirá un mensaje de error e instrucciones para iniciar sesión primero antes continuo. En la página de perfil, se realiza una verificación de las cookies de su navegador para vea si su cookie se configuró en la página de inicio de sesión.

Realizar un seguimiento de las cookies es fácil con la biblioteca de Solicitudes:

````python
params = {
    "username": "rya",
    "password": "password",
}

response = requests.post(
    "https://pythonscraping.com/pages/cookies/welcome.php",
    params,
)

print(f"Cookie is set to:\n{response.cookies.get_dict()}")

print("Going to profile page...")

response = requests.get(
    "https://pythonscraping.com/pages/cookies/profile.php",
    cookies=response.cookies,
)

print(response.text)
````

Lo que acabamos de hacer es hacer una solicitud mediante formulario, obtener las cookies y entrar a la otra página web junto con las cookies

El script anterior funciona en página simples que requieran aceptar las cookies pero en casos más complejos en los que se cambien las cookies contantemente o en los que se inicie sesión de manera constante lo más conveniente es lo siguiente:

````python
response = session.post(
    "https://pythonscraping.com/pages/cookies/welcome.php",
    params,
)

print(f"Cookie is set to:\n{response.cookies.get_dict()}")

print("Going to profile page...")

response = session.get(
    "https://pythonscraping.com/pages/cookies/profile.php",
)

print(response.text)
````

En este caso el objeto Session() lo que hace es seguir el rastro de la información de la sesión como las cookies, los encabezados y la información sobre los protocolos.

#### Medidas con autenticación auth
Antes de que salga la ventana de aceptar las cookies es posible que cuando quieras acceder a ciertos sitios te exigan una autenticación esto sucede en páginas las cuales tienen mucha seguridad. La página http://pythonscraping.com/pages/auth/login.php es una:

Una manera de poder acceder es de la manera siguiente:
session

````python
import requests
from requests.auth import HTTPBasicAuth

auth = HTTPBasicAuth("ryan", "password")
response = requests.post(
    "https://pythonscraping.com/pages/auth/login.php",
    auth=auth,
)

print(response.text)
````
